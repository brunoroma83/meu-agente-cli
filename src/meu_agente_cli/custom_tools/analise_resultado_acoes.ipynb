```python
import pandas as pd

arquivo = "estudos_acoes/2026-08-26_acoes.csv"

df = pd.read_csv(arquivo)

df.columns
```

```python
# mostrar os maiores retornos calculdos

df_top10 = df.sort_values(by='retorno_esperado_pct', ascending=False).drop(columns=['half_life_semanas_configurada']).head(10)
print(df_top10)
```

```python
df_filtrado = (
    df[df['mae_validacao_pct'] < 10]
    .sort_values(by='retorno_esperado_pct', ascending=False).drop(columns=['half_life_semanas_configurada']).head(10)
)

print(df_filtrado)
```
